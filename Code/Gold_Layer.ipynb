```python
%sql
create or replace view hospital_db.hospital.total_num_patients_gold
as
 SELECT 
      COUNT(DISTINCT patient_sk) AS num_patients
FROM hospital_db.hospital.fact_table_silver
```

```python
%sql
create or replace view hospital_db.hospital.avg_stay_hours_gold
as
SELECT 
    round(AVG(length_of_stay_hours),0) AS avg_los_hours
FROM hospital_db.hospital.fact_table_silver
```

```python
%sql
SELECT 
    concat('Hopsital Num ', hospital_id) AS Hopsital 
FROM hospital_db.hospital.fact_table_silver 
group by hospital_id
ORDER BY  COUNT(DISTINCT patient_sk) desc
LIMIT 1
```

```python
%sql
create or replace view hospital_db.hospital.top_hospital_gold
as
SELECT 
    concat('Hopsital Num ', hospital_id) AS Hopsital
FROM hospital_db.hospital.fact_table_silver 
group by hospital_id
ORDER BY  COUNT(DISTINCT patient_sk) desc
LIMIT 1
```

```python
%sql
create or replace view hospital_db.hospital.top_department_gold as
SELECT 
    d.department
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_department_silver d
  ON f.department_sk = d.department_sk
GROUP BY d.department
ORDER BY  COUNT(DISTINCT f.patient_sk) DESC
LIMIT 1;
```

```python
%sql
create or replace view hospital_db.hospital.fact_table_silverdf_Num_patients_by_department_gold as 
SELECT 
    d.department,
    COUNT(DISTINCT f.patient_sk) AS num_patients
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_department_silver d ON f.department_sk = d.department_sk
GROUP BY d.department
ORDER BY num_patients DESC
```

```python
%sql
create or replace view hospital_db.hospital.length_of_stay_hours_gold
as 
SELECT 
    d.department,
    round(AVG(f.length_of_stay_hours),0) AS avg_los_hours
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_department_silver d ON f.department_sk = d.department_sk
GROUP BY d.department
ORDER BY avg_los_hours DESC
```

```python
%sql
create or replace view hospital_db.hospital.Num_patients_by_gender_gold
as
SELECT 
    p.gender,
    COUNT(DISTINCT f.patient_sk) AS num_patients
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_patient_silver p ON f.patient_sk = p.patient_sk
GROUP BY p.gender
ORDER BY num_patients DESC
```

```python
%sql
create or replace view hospital_db.hospital.Num_patients_by_age_gold
as
SELECT 
    p.age,
    COUNT(DISTINCT f.patient_sk) AS num_patients
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_patient_silver p ON f.patient_sk = p.patient_sk
GROUP BY p.age
ORDER BY p.age
```

```python
%sql
create or replace view hospital_db.hospital.daily_patients_gold
as
SELECT 
    d.date_sk,
    COUNT(DISTINCT f.patient_sk) AS num_admissions
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_date_silver d ON f.admission_date_sk = d.date_sk
GROUP BY d.date_sk
ORDER BY d.date_sk
```

```python
%sql
create or replace view hospital_db.hospital.current_active_patients_gold
as
SELECT 
    COUNT(DISTINCT patient_sk) AS active_patients
FROM hospital_db.hospital.fact_table_silver 
WHERE 
    discharge_date_sk IS NULL 
```

```python
%sql
create or replace view hospital_db.hospital.Num_patients_by_age_group_gold
as
SELECT 
    CASE
        WHEN p.age <= 18 THEN '0-18 (Children)'
        WHEN p.age > 18 AND p.age <= 35 THEN '19-35 (Youth)'
        WHEN p.age > 35 AND p.age <= 55 THEN '36-55 (Adults)'
        ELSE '56+ (Seniors)'
    END AS age_group,
    COUNT(DISTINCT f.patient_sk) AS num_patients
FROM hospital_db.hospital.fact_table_silver f
JOIN hospital_db.hospital.dim_patient_silver p ON f.patient_sk = p.patient_sk
GROUP BY age_group
ORDER BY num_patients DESC
```

```python
%sql
create or replace view hospital_db.hospital.hospital_performance_gold
as
SELECT 
    concat('Hopsital Num ', hospital_id) AS Hospital,
    COUNT(DISTINCT patient_sk) AS num_patients,
    round(AVG(length_of_stay_hours), 0) AS avg_los_hours
FROM hospital_db.hospital.fact_table_silver 
GROUP BY hospital_id
ORDER BY num_patients DESC
```
