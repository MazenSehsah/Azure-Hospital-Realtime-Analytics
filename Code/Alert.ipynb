```python
import requests
 
logic_app_url = "https://prod-21.uaenorth.logic.azure.com:443/workflows/914331b3e7924021bdaab67983c5eaa8/triggers/When_an_HTTP_request_is_received/paths/invoke?api-version=2016-10-01&sp=%2Ftriggers%2FWhen_an_HTTP_request_is_received%2Frun&sv=1.0&sig=TyCBHltgheDQJa7eTVZWlC7gkyWh6vaCNUZDALRp2Fw"
 
num_patients = spark.sql("""
SELECT COUNT(DISTINCT patient_sk) AS num_patients_last_hour
FROM hospital_db.hospital.fact_table_silver
WHERE admission_time >= current_timestamp() - INTERVAL 1 HOURS
""").collect()[0]["num_patients_last_hour"]
 
if num_patients >= 0:
    response = requests.post(
        logic_app_url,
        json={"num_patients_last_hour": num_patients}
    )
    print("Alert sent!", response.status_code)
else:
    print("No alert needed. Patients:", num_patients)
```
